# My Custom Questions (Day 2 Task)

Two questions of my own about the Titanic dataset, answered with Pandas filtering (`loc`, `value_counts`, `mean`).

## Question 1: Did women in 3rd class survive as often as women in 1st and 2nd class?

In class we saw that most women survived. I wanted to check whether that was true for **every** class, or only for the richer ones.

```python
# Women in 3rd class
q1 = df.loc[(df["Sex"] == "female") & (df["Pclass"] == 3)]

print("Women in 3rd class:", len(q1))
print("\nSurvived (1) vs Did not survive (0):")
display(q1["Survived"].value_counts())
print("Survival rate:", round(q1["Survived"].mean() * 100, 1), "%")
```

```python
# Compare with the other classes
for pclass in [1, 2, 3]:
    women = df.loc[(df["Sex"] == "female") & (df["Pclass"] == pclass)]
    rate = women["Survived"].mean() * 100
    print(f"Class {pclass}: {len(women)} women, survival rate = {round(rate, 1)}%")
```

**Finding:** Only **50%** of women in 3rd class survived (72 out of 144). In 1st class almost every woman survived (**96.8%**), and in 2nd class **92.1%** did. So "women first" mostly helped women in the upper classes. A woman in 3rd class had about the same odds as a coin toss.

## Question 2: How many children under 12 travelling with a parent survived, and where did they board?

Children travelling with a parent have `Parch > 0`. I wanted to see how many of them survived, which port the survivors came from, and whether their class mattered.

```python
# Children under 12 travelling with at least one parent
kids = df.loc[(df["Age"] < 12) & (df["Parch"] > 0)]
surviving_kids = kids.loc[kids["Survived"] == 1]

print("Children under 12 with a parent:", len(kids))
print("Survived:", len(surviving_kids))
print("Survival rate:", round(kids["Survived"].mean() * 100, 1), "%")

print("\nWhere the surviving children boarded (S = Southampton, C = Cherbourg, Q = Queenstown):")
display(surviving_kids["Embarked"].value_counts())
```

```python
# Did the children's class matter?
for pclass in [1, 2, 3]:
    group = kids.loc[kids["Pclass"] == pclass]
    rate = group["Survived"].mean() * 100
    print(f"Class {pclass}: {len(group)} children, survival rate = {round(rate, 1)}%")
```

**Finding:** There were **66** children under 12 travelling with a parent, and **38** of them survived (**57.6%**). Most survivors boarded at **Southampton (30)**, the rest at **Cherbourg (8)**, and none from Queenstown. Class made a big difference: **all 17** children in 2nd class survived, but only **40%** of the 45 children in 3rd class did.

**Overall:** Both questions show the same pattern. Being a woman or a child helped, but passenger class still decided a lot about who survived.
